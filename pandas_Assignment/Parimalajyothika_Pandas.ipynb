### Q1. NumPy Broadcasting

**Question:**

Two NumPy arrays have shapes `(3, 1)` and `(1, 3)`. Predict the shape of their element-wise sum and explain how broadcasting works. Why might the result differ from adding two one-dimensional arrays?

````python
import numpy as np

a = np.array([[10], [20], [30]])
b = np.array([[1, 2, 3]])

result = a + b

print(result)
print(result.shape)
````

### Predicted Output

```text
[[11 12 13]
 [21 22 23]
 [31 32 33]]
(3, 3)
```

### Shapes of Arrays

The shape of `a` is **(3, 1)** and the shape of `b` is **(1, 3)**.

### Broadcasting

Broadcasting allows NumPy to perform element-wise operations on arrays with compatible shapes. Here, the dimensions with size `1` are expanded to match the other array. Therefore, the arrays are broadcast to a common shape of **(3, 3)** before addition.

### If `b` is replaced with `np.array([1, 2])`

If `b` is replaced with `np.array([1, 2])`, its shape is **(2,)**. It can be broadcast with `a`, which has shape **(3, 1)**.

The result will have shape **(3, 2)**:

```text
[[11 12]
 [21 22]
 [31 32]]
```

The result is different because the new `b` contains two values, so the resulting array has two columns instead of three.

### Explanation

Broadcasting makes it possible to add arrays with compatible shapes without manually changing their dimensions. Thus, `(3, 1)` and `(1, 3)` produce a `(3, 3)` result, while `(3, 1)` and `(2,)` produce a `(3, 2)` result.

### Q2. CSV Parsing, Missing Values and Dates

**Question:**

A CSV file contains commas inside quoted text fields, missing values, and dates in different formats. Explain how incorrect parsing can affect the resulting DataFrame and identify suitable read options or preprocessing steps.

````python
import pandas as pd

df = pd.read_csv("students.csv")

print(df)
print(df.dtypes)
````

### Quoted Comma

The comma in `"Asha, R"` is inside quotation marks, so Pandas treats it as part of the `Name` value instead of creating an extra column.

### Issues with Date and Marks

The `Date` column contains different date formats, so it should be converted into a consistent date-time format. The `Marks` column contains a blank value and `NA`, which need to be investigated before treating them as missing values.

### Corrected Code

````python
import pandas as pd
import numpy as np

df = pd.read_csv("students.csv", keep_default_na=False)

print("Unique Marks values:", df["Marks"].unique())

df["Date"] = pd.to_datetime(
    df["Date"],
    format="mixed",
    errors="coerce",
    dayfirst=True
)

df["Marks"] = df["Marks"].replace("", np.nan)

print(df)
print(df.dtypes)
````

### Explanation

The quoted comma is correctly handled by `read_csv()`, the dates are converted to a common date-time format, and the marks are inspected so that `NA` is not treated as missing without first understanding its meaning.

### Q1. Student Assessment Data Cleaning

**Question:**

A student assessment dataset contains missing marks, duplicate student records, inconsistent department names such as IT, it, and I.T., and marks stored as text. Develop a cleaning sequence to standardize the data, handle missing values, remove duplicates, and convert marks into a suitable numeric type. State one risk of deleting missing values without investigating why they are missing.

### Dataset

The dataset contains student assessment records with inconsistent department names, duplicate records, missing marks, text values in the Marks column, and an invalid mark outside the range of 0–100.

````python
import pandas as pd
import numpy as np

df = pd.read_csv("student_assessment_dirty.csv", sep="\t", keep_default_na=True)

print(df.head())
df.info()
print(df.isnull().sum())
````

### Standardize Department Names

The department names are standardized so that different forms such as `IT`, `it`, and `I.T.` are represented consistently, and the same approach is applied to `Finance`.

````python
df["Department"] = df["Department"].replace({
    "IT": "IT",
    "it": "IT",
    "I.T.": "IT",
    "Finance": "Finance",
    "finance": "Finance"
})

print(df["Department"].value_counts())
````

### Remove Duplicate Records

The dataset contains an exact duplicate student assessment record, so duplicate rows are identified and removed.

````python
duplicates_removed = df.duplicated().sum()

df = df.drop_duplicates()

print("Duplicate records removed:", duplicates_removed)
````

### Convert Marks to Numeric

The Marks column is first checked to distinguish `Absent`, `NA`, and blank entries. Numeric values and values such as `88 marks` are then converted into numeric form.

````python
df["Mark_Status"] = df["Marks"].astype("string").str.strip()

df["Mark_Status"] = df["Mark_Status"].replace({
    "Absent": "Absent"
}).fillna("Not Entered")

df["Marks"] = pd.to_numeric(
    df["Marks"].astype("string").str.extract(
        r"(\d+(?:\.\d+)?)",
        expand=False
    ),
    errors="coerce"
)
````

### Handle Missing Marks

Missing marks should be investigated before deleting them. `Absent` and a mark that has not yet been entered may have different meanings, so they should not automatically be treated in the same way.

One risk of deleting missing values without investigating them is that valid information may be lost and the analysis may become biased.

### Identify Invalid Marks

Marks outside the valid range of 0–100 are identified before deciding how to handle them.

````python
invalid_mask = df["Marks"].notna() & ~df["Marks"].between(0, 100)

invalid_marks = df[invalid_mask]

print("Invalid marks:")
print(invalid_marks)

df.loc[invalid_mask, "Mark_Status"] = "Invalid"
df.loc[invalid_mask, "Marks"] = np.nan
````

### Cleaned DataFrame and Summary

The cleaned DataFrame is displayed along with the number of duplicate records removed, missing marks remaining, and invalid marks identified.

````python
print("Cleaned DataFrame:")
print(df)

print("\nRecords removed:", duplicates_removed)
print(
    "Missing marks remaining:",
    df["Mark_Status"].isin(["Absent", "Not Entered"]).sum()
)
print("Invalid marks identified:", len(invalid_marks))
````

### Explanation

The dataset is cleaned by standardizing departments, removing duplicates, distinguishing missing and absent marks, converting marks to numeric values, and identifying invalid marks.
